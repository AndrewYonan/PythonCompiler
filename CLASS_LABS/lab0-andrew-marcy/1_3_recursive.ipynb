```python
import ast
from ast import *
tree = ast.parse("""
x5 = 5
x6 = 10
print(x5+x6)
""")
print(ast.dump(tree,indent=2))
```

```python
#print names using recursion
def print_names(n):
    if (isinstance(n, Module)):
        for x in n.body:
            print_names(x)
    if (isinstance(n, Assign)):
        for x in n.targets:
            print(x.id)
        print_names(n.value)
    if (isinstance(n, Expression)):
        print_names(n.value)
    if (isinstance(n, BinOp)):
        print_names(n.left)
        print_names(n.right)
    if (isinstance(n, UnaryOp)):
        print_names(n.operand)
    if(isinstance(n, Call)):
       print_names(n.func)
       print_names(n.args)
       
```

```python
print_names(tree)
```
